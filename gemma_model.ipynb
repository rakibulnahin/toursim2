```python
import json
from llama_cpp import Llama

class MyModel():


    # 1. Initialize and automatically download the GGUF model from Hugging Face
    # n_gpu_layers=0 forces CPU execution
    # n_threads uses multi-threading (adjust to match your physical CPU core count)
    
    llm = Llama.from_pretrained(
        repo_id="ggml-org/gemma-4-E2B-it-GGUF",
        filename="gemma-4-E2B-it-Q4_0.gguf",  # Exact filename without wildcard
        n_gpu_layers=0,
        n_threads=4,
        n_ctx=2048,
        verbose=False,
    )


    # 3. Request inference with JSON mode enforcement
    def extract(self, system_prompt:str, user_input:str ) -> str:
        response = self.llm.create_chat_completion(
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_input},
            ],
            response_format={"type": "json_object"},
            temperature=0.1,  # Low temperature for stable structure
        )

        # 4. Extract and print the parsed result
        raw_output = response["choices"][0]["message"]["content"]
        data = json.loads(raw_output)
        return data

    



```
